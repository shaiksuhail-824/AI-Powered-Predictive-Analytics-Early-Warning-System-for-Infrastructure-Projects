# 04 — Feature Engineering & CUF Baseline Comparison

Compares the Common Utility Format (CUF) baseline features against the full extended feature matrix, confirming predictive lift potential and data leakage prevention.


```python
import pandas as pd

cuf_df = pd.read_csv('../data/features/paimana_cuf_baseline.csv')
ml_df = pd.read_csv('../data/features/paimana_ml_ready_time_overrun.csv')

print(f'CUF Baseline Features: {cuf_df.shape[1]} attributes')
print(f'Full Engineered ML-Ready Features: {ml_df.shape[1]} attributes')
```

### Correlation of Engineered Features with Overrun Target


```python
corr = ml_df.corr(numeric_only=True)['time_overrun_flag'].sort_values()
print('Top 5 Inversely Correlated Features with Delay:')
print(corr.head(5))
print('\nTop 5 Positively Correlated Features with Delay:')
print(corr.tail(6))
```

### Anti-Leakage Verification


```python
forbidden = ['revised_cost_cr', 'revised_doc_dt', 'actual_completion_date']
for col in forbidden:
    assert col not in ml_df.columns, f'Data leakage alert: {col} present in feature matrix!'
print('Verification passed: Zero future information or target leakage detected.')
```
