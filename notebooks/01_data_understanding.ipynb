This notebook inspects the raw PAIMANA Time Overrun dataset ingested into `data/raw/paimana_time_overrun.csv`.


```python
import pandas as pd
import numpy as np

raw_path = '../data/raw/paimana_time_overrun.csv'
df = pd.read_csv(raw_path, low_memory=False)
print(f'Raw Dataset Shape: {df.shape}')
df.head()
```

### Column Inventory & Missing Value Analysis


```python
missing_df = pd.DataFrame({
    'Data Type': df.dtypes,
    'Missing Count': df.isna().sum(),
    'Missing Pct (%)': (df.isna().sum() / len(df) * 100).round(2)
})
missing_df
```
