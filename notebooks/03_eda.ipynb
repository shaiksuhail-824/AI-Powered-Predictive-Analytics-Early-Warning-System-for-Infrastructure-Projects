# 03 — Exploratory Data Analysis (EDA)

Visualizes infrastructure project delay distributions, physical progress vs expenditure parity, schedule-progress gaps, and temporal delay trajectories.


```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv('../data/interim/paimana_time_overrun_validated.csv', low_memory=False)
plt.style.use('seaborn-v0_8-whitegrid')
```

### Target Overrun Distribution


```python
fig, ax = plt.subplots(figsize=(6, 4))
sns.countplot(data=df, x='time_overrun_flag', palette=['green', 'red'], ax=ax)
ax.set_title('Schedule Overrun Classification (0=On-Time, 1=Delayed)')
plt.show()
```

### Physical Progress vs Expenditure


```python
fig, ax = plt.subplots(figsize=(8, 6))
ax.scatter(df['physical_progress_pct'], df['expenditure_to_original_cost_pct'].clip(upper=200), c=df['time_overrun_flag'], cmap='coolwarm', alpha=0.3, s=15)
ax.plot([0, 100], [0, 100], 'k--', label='Parity Line')
ax.set_xlabel('Physical Progress (%)')
ax.set_ylabel('Expenditure / Original Cost (%)')
ax.legend()
plt.show()
```

### Schedule-Progress Gap as Early Warning Discriminator


```python
fig, ax = plt.subplots(figsize=(7, 4))
sns.boxplot(data=df, x='time_overrun_flag', y=df['schedule_progress_gap_pct'].clip(-150, 100), palette=['green', 'red'], ax=ax)
ax.set_title('Schedule-Progress Gap by Overrun Status')
plt.show()
```
