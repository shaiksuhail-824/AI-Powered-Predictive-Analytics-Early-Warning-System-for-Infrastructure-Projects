**SIH26103 — Infrastructure Project Monitoring**

Validates data integrity, non-negative financial constraints, physical progress bounds [0, 100]%, duplicate detection, and missing indicator flags.


```python
import pandas as pd
import numpy as np

interim_path = '../data/interim/paimana_time_overrun_validated.csv'
df = pd.read_csv(interim_path, low_memory=False)
print(f'Validated Dataset Shape: {df.shape}')
```

### Check 1: Record & Key Uniqueness


```python
duplicates = df.duplicated().sum()
key_duplicates = df.duplicated(subset=['project_code', 'report_year', 'report_month_num']).sum()
print(f'Exact duplicate rows: {duplicates}')
print(f'Duplicate (project, year, month) records: {key_duplicates}')
```

### Check 2: Physical Progress & Financial Bounds


```python
assert (df['original_cost_cr'] >= 0).all(), 'Negative cost found!'
assert (df['cumulative_expenditure_cr'] >= 0).all(), 'Negative spend found!'
assert ((df['physical_progress_pct'] >= 0) & (df['physical_progress_pct'] <= 100)).all(), 'Progress out of bounds!'
print('All invariant checks passed: Non-negative costs and bounded progress.')
```

### Check 3: Ground-Truth Target Consistency


```python
print(df['time_overrun_flag'].value_counts(dropna=False))
print(f'Delay Rate: {(df["time_overrun_flag"] == 1).mean() * 100:.2f}%')
```
